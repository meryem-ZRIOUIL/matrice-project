```python
import pandas as pd

data = {
    "minute":[1,2,3,4,5],
    "vehicules":[120,150,200,180,160],
    "tweets":[15,20,50,40,30],
    "pluie":["Non","Non","Oui","Oui","Non"]
}

df = pd.DataFrame(data)


df["moyenne_mobile"] = df["vehicules"].rolling(window=3).mean()

temps_base = 10

df["temps_trajet"] = df["pluie"].apply(
    lambda x: temps_base*1.2 if x=="Oui" else temps_base
)

print(df)
```
